# How to make a mini-Qiskit from scratch with TensorKrowch

We are going to build a small Qiskit-inspired circuit class on top of TensorKrowch, using **PyTorch as the only tensor backend**. The goal is pedagogical: keep the implementation readable, support GPU when available, and make the tensor network ideas explicit.

**What we build:**
- **TNCircuit**: Low-level class that stores tensor layers and performs contractions (statevector or MPS).
- **QuantumRegister**: Maps qubit positions to labels (e.g. `q(0)`, `q(1)`).
- **QuantumCircuit**: Qiskit-like API (`.h()`, `.cx()`, `.ccx()`, etc.) on top of TNCircuit.

**Key features:**
- Two contraction modes: **statevector** (exact dense state) and **MPS** (with optional TT-style compression).
- Controlled gates use auxiliary "control-signal" tensors that propagate along the circuit.
- All queries (`state_vector`, `expected`, `check_state`) work on copies of the network, so the circuit is never mutated.


## Packages

We use:

- **`torch`**: Tensors, complex arithmetic, optional GPU execution, and the SVD used for MPS truncation.
- **`tensorkrowch`**: Explicit nodes and edges on top of PyTorch tensors. We use it to wire (`in`, `out`, `up`, `down`) and contract the circuit graph.
- **`matplotlib`**: Circuit visualization (draws gates and control wires).
- **`numpy`**: Only for friendly outputs in demos (e.g. `state_vector`), not as an execution backend.


## Axis conventions

The circuit uses four axis labels:

- **`in`**: Input physical index of a gate. Connects to the previous layer's `out` on the same qudit.
- **`out`**: Output physical index of a state or gate. Carries the state forward in time.
- **`up`** and **`down`**: Virtual bonds for multi-qubit gates. Control wires use these to propagate a "control signal" (0 or 1) between qubits. In MPS, they are the left/right bond dimensions.

For the MPS representation we keep a canonical core order `[up, out, down]`. That matches the helper's TT-core format and keeps the GPU path entirely in PyTorch.


```python
# Imported modules and runtime configuration
import numpy as np
import tensorkrowch as tk
import torch

import matplotlib.pyplot as plt
from matplotlib import rcParams

# Auxiliary functions: tensor conversion, grid management, TensorKrowch graph
# helpers, MPS canonicalization, control tensors, and TT-style compression.
from auxiliary_functions_tensorkrowch import (
    TT_State_Compression,
    append_operator_to_grid,
    build_adjoint_grid,
    build_basis_bra_layer,
    build_control_tensors,
    build_controlled_target_tensor,
    build_multi_control_tensors,
    canonicalize_mps_node,
    connect_circuit_layers,
    connect_edges,
    contract_between,
    contract_grid_row,
    copy_node_grid,
    ensure_torch_tensor,
    gate_style,
    mps_nodes_to_dense_tensor,
    mps_nodes_to_scalar,
    process_axes,
    reorder_dense_output_axes,
    tensor_to_numpy_flat,
    tensor_to_python_scalar,
)

# Use complex128 for full precision; device auto-selects GPU if available
DTYPE = torch.complex128
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Plotting style for circuit diagrams
plt.style.use("dark_background")
rcParams.update({"xtick.labelsize": 25, "ytick.labelsize": 25, "font.size": 30})

print(f"Running with PyTorch on: {device}")
```

# Runtime helpers

Before defining the circuit classes we collect a few small helpers:

- **`make_node`**: Wraps a tensor as a TensorKrowch node with the given axis names. Ensures the tensor is on the correct device and dtype.
- **Standard gates**: X, Y, Z, H as constant tensors; RX, RY, RZ as parameterized gates (angle `theta`).


```python
# Tensor and gate helpers
def make_node(tensor, name, axis_names):
    """Create a TensorKrowch node with the given tensor, name, and axis labels."""
    tensor = ensure_torch_tensor(tensor, device=device, dtype=DTYPE)
    return tk.Node(
        shape=tensor.shape,
        axes_names=axis_names,
        name=name,
        tensor=tensor,
    )


# Pauli and Hadamard gates (2x2 matrices in computational basis)
X_GATE = ensure_torch_tensor([[0, 1], [1, 0]], device=device, dtype=DTYPE)
Y_GATE = ensure_torch_tensor([[0, -1j], [1j, 0]], device=device, dtype=DTYPE)
Z_GATE = ensure_torch_tensor([[1, 0], [0, -1]], device=device, dtype=DTYPE)
H_GATE = ensure_torch_tensor([[1, 1], [1, -1]], device=device, dtype=DTYPE)
H_GATE = H_GATE / torch.sqrt(torch.tensor(2.0, dtype=torch.float64, device=device))


def rx_gate(theta):
    theta_value = float(theta)
    theta_tensor = torch.as_tensor(theta_value, dtype=torch.float64, device=device)
    cos_half = torch.cos(theta_tensor / 2)
    sin_half = torch.sin(theta_tensor / 2)
    tensor = torch.stack(
        [
            torch.stack([cos_half, -1j * sin_half]),
            torch.stack([-1j * sin_half, cos_half]),
        ]
    ).to(DTYPE)
    return tensor, f"RX({theta_value:.3f})"


def ry_gate(theta):
    theta_value = float(theta)
    theta_tensor = torch.as_tensor(theta_value, dtype=torch.float64, device=device)
    cos_half = torch.cos(theta_tensor / 2)
    sin_half = torch.sin(theta_tensor / 2)
    tensor = torch.stack(
        [
            torch.stack([cos_half, -sin_half]),
            torch.stack([sin_half, cos_half]),
        ]
    ).to(DTYPE)
    return tensor, f"RY({theta_value:.3f})"


def rz_gate(theta):
    theta_value = float(theta)
    theta_tensor = torch.as_tensor(theta_value, dtype=torch.float64, device=device)
    tensor = torch.stack(
        [
            torch.stack([
                torch.exp(-0.5j * theta_tensor),
                torch.zeros((), dtype=DTYPE, device=device),
            ]),
            torch.stack([
                torch.zeros((), dtype=DTYPE, device=device),
                torch.exp(0.5j * theta_tensor),
            ]),
        ]
    ).to(DTYPE)
    return tensor, f"RZ({theta_value:.3f})"
```

# Query and visualization helpers

- **`contraction_to_scalar`**: Extracts a single complex number from the contraction result. Handles both statevector (single-node result) and MPS (chain of nodes) representations.
- **`draw_circuit`**: Plots the circuit as a grid: horizontal lines = qubits, markers = gates. Control wires (up/down) are drawn as vertical segments.


```python
# Query and visualization helpers
def contraction_to_scalar(result, representation: str):
    """Extract a Python scalar from a contraction result (statevector or MPS)."""
    if representation.lower() == "mps":
        return mps_nodes_to_scalar(result)
    return tensor_to_python_scalar(result.tensor.reshape(-1)[0])


def draw_circuit(node_grid, depth: int, n_qudits: int) -> str:
    """Plot the circuit: qubits as horizontal lines, gates as boxes, control wires as vertical segments."""
    from matplotlib.patches import FancyBboxPatch, Circle

    wire_spacing = 1.2
    gate_width = 0.7
    gate_height = 0.5

    fig, ax = plt.subplots(
        figsize=(max(depth * 0.8 + 3, 6), max(n_qudits * wire_spacing + 2, 4)),
        facecolor="#1a1a2e",
    )
    ax.set_facecolor("#1a1a2e")
    ax.set_xlim(-0.8, depth + 0.3)
    ax.set_ylim(-0.6, n_qudits * wire_spacing - 0.4)
    ax.set_aspect("equal")
    ax.axis("off")

    # Qubit wires (horizontal lines)
    wire_color = "#4a5568"
    wire_lw = 1.5
    for qudit in range(n_qudits):
        y_pos = (n_qudits - 1 - qudit) * wire_spacing
        ax.plot([-0.3, depth + 0.1], [y_pos, y_pos], color=wire_color, lw=wire_lw, zorder=0)
        ax.text(-0.65, y_pos, f"q{qudit}", fontsize=9, color="#94a3b8", va="center", ha="right")

    # First pass: draw control wires (vertical segments) behind gates
    ctrl_color = "#4a9eff"
    ctrl_lw = 2.0
    for layer in range(depth):
        for qudit in range(n_qudits):
            tensor = node_grid[layer][qudit]
            if tensor is None:
                continue
            y_pos = (n_qudits - 1 - qudit) * wire_spacing
            if "up" in tensor.axes_names:
                ax.plot([layer, layer], [y_pos, y_pos + wire_spacing], color=ctrl_color, lw=ctrl_lw, zorder=1)
            if "down" in tensor.axes_names:
                ax.plot([layer, layer], [y_pos, y_pos - wire_spacing], color=ctrl_color, lw=ctrl_lw, zorder=1)

    # Second pass: draw gates
    for layer in range(depth):
        for qudit in range(n_qudits):
            tensor = node_grid[layer][qudit]
            if tensor is None:
                continue

            y_pos = (n_qudits - 1 - qudit) * wire_spacing
            gate_name = tensor.name
            facecolor, edgecolor, label = gate_style(gate_name)

            if gate_name == "Pass":
                circle = Circle((layer, y_pos), 0.08, facecolor=facecolor, edgecolor=edgecolor, zorder=2)
                ax.add_patch(circle)
                continue

            is_control = "Ctrl_" in gate_name or "cCtrl_" in gate_name
            if is_control:
                circle = Circle(
                    (layer, y_pos), 0.2,
                    facecolor=facecolor, edgecolor=edgecolor, linewidth=1.5, zorder=3
                )
                ax.add_patch(circle)
                ax.text(layer, y_pos, label, fontsize=10, color="#0f172a", va="center", ha="center", zorder=4)
            else:
                box = FancyBboxPatch(
                    (layer - gate_width / 2, y_pos - gate_height / 2),
                    gate_width, gate_height,
                    boxstyle="round,pad=0.02,rounding_size=0.08",
                    facecolor=facecolor, edgecolor=edgecolor, linewidth=1.5, zorder=3
                )
                ax.add_patch(box)
                display_label = label if label else gate_name
                if len(display_label) > 10:
                    mid = len(display_label) // 2
                    display_label = f"{display_label[:mid]}\n{display_label[mid:]}"
                ax.text(
                    layer, y_pos, display_label,
                    fontsize=8, va="center", ha="center", color="#0f172a",
                    zorder=4
                )

    plt.tight_layout()
    plt.show()
    return ""
```

# Controlled-gate helpers

Multi-qubit gates (CNOT, CZ, Toffoli) are built from:
1. **Control tensors** (Ctrl_Up, Ctrl_Down, Pass): Propagate a 0/1 signal along the circuit.
2. **Controlled target tensor**: Applies the base gate only when the control signal is 1.

**`build_control_gate_layers`**: For a list of (control, target) pairs, builds the tensor layout. If control is above target, the signal goes "down"; if below, it goes "up". Pass tensors relay the signal through intermediate qubits.

**`build_multi_control_gate_layer`**: For Toffoli (multiple controls, one target), uses cctrl_up/cctrl_down tensors when the target lies between controls.


```python
# Controlled-gate layout helpers
def build_control_gate_layers(base_tensor, name: str, positions_control: list[int], positions_target: list[int]):
    """Build tensor layers for single-control gates (e.g. CNOT, CZ) with arbitrary control/target pairs."""
    if len(positions_control) != len(positions_target):
        raise ValueError("Control and target lists must have the same length.")
    if not positions_control:
        raise ValueError("At least one controlled gate must be specified.")

    min_qudit = min(min(positions_control), min(positions_target))
    max_qudit = max(max(positions_control), max(positions_target))
    total_qudits = max_qudit - min_qudit + 1
    positions = list(range(min_qudit, max_qudit + 1))

    ctrl_tensor_down, ctrl_tensor_up, pass_tensor = build_control_tensors(
        device=device, dtype=DTYPE
    )
    controlled_tensor = build_controlled_target_tensor(
        base_tensor, device=device, dtype=DTYPE, n_signals=1
    )

    total_layers = [[None for _ in range(total_qudits)] for _ in range(len(positions_control))]

    for gate_index, (control, target) in enumerate(zip(positions_control, positions_target)):
        if control == target:
            raise ValueError("Control and target qudits must be different.")

        difference = control - target
        if difference > 0:
            total_layers[gate_index][target - min_qudit] = make_node(
                controlled_tensor, name, ["in", "out", "down"]
            )
            total_layers[gate_index][control - min_qudit] = make_node(
                ctrl_tensor_up, "Ctrl_Up", ["in", "out", "up"]
            )
            for intermediate in range(1, difference):
                total_layers[gate_index][target + intermediate - min_qudit] = make_node(
                    pass_tensor, "Pass", ["in", "out", "up", "down"]
                )
        else:
            total_layers[gate_index][target - min_qudit] = make_node(
                controlled_tensor, name, ["in", "out", "up"]
            )
            total_layers[gate_index][control - min_qudit] = make_node(
                ctrl_tensor_down, "Ctrl_Down", ["in", "out", "down"]
            )
            for intermediate in range(1, -difference):
                total_layers[gate_index][target - intermediate - min_qudit] = make_node(
                    pass_tensor, "Pass", ["in", "out", "up", "down"]
                )

    return positions, total_layers


def build_multi_control_gate_layer(base_tensor, name: str, positions_control: list[int], position_target: int):
    """Build tensor layers for multi-control gates (e.g. Toffoli) with one target qubit."""
    if not positions_control:
        raise ValueError("At least one control qudit is required.")
    if position_target in positions_control:
        raise ValueError("The target qudit cannot also be a control qudit.")

    min_qudit = min(min(positions_control), position_target)
    max_qudit = max(max(positions_control), position_target)
    total_qudits = max_qudit - min_qudit + 1
    positions = list(range(min_qudit, max_qudit + 1))

    tensors = build_multi_control_tensors(device=device, dtype=DTYPE)
    total_layers = [[
        make_node(tensors["pass"], "Pass", ["in", "out", "up", "down"])
        for _ in range(total_qudits)
    ]]

    if min_qudit == position_target:
        controlled_tensor = build_controlled_target_tensor(
            base_tensor, device=device, dtype=DTYPE, n_signals=1
        )
        total_layers[0][position_target - min_qudit] = make_node(
            controlled_tensor, name, ["in", "out", "down"]
        )
        for control in positions_control:
            idx = control - min_qudit
            if control == max_qudit:
                total_layers[0][idx] = make_node(
                    tensors["ctrl_up"], "Ctrl_Up", ["in", "out", "up"]
                )
            else:
                total_layers[0][idx] = make_node(
                    tensors["cctrl_up"], "cCtrl_Up", ["in", "out", "up", "down"]
                )
    elif max_qudit == position_target:
        controlled_tensor = build_controlled_target_tensor(
            base_tensor, device=device, dtype=DTYPE, n_signals=1
        )
        total_layers[0][position_target - min_qudit] = make_node(
            controlled_tensor, name, ["in", "out", "up"]
        )
        for control in positions_control:
            idx = control - min_qudit
            if control == min_qudit:
                total_layers[0][idx] = make_node(
                    tensors["ctrl_down"], "Ctrl_Down", ["in", "out", "down"]
                )
            else:
                total_layers[0][idx] = make_node(
                    tensors["cctrl_down"], "cCtrl_Down", ["in", "out", "up", "down"]
                )
    else:
        controlled_tensor = build_controlled_target_tensor(
            base_tensor, device=device, dtype=DTYPE, n_signals=2
        )
        total_layers[0][position_target - min_qudit] = make_node(
            controlled_tensor, name, ["in", "out", "up", "down"]
        )
        for control in positions_control:
            idx = control - min_qudit
            if control == min_qudit:
                total_layers[0][idx] = make_node(
                    tensors["ctrl_down"], "Ctrl_Down", ["in", "out", "down"]
                )
            elif control == max_qudit:
                total_layers[0][idx] = make_node(
                    tensors["ctrl_up"], "Ctrl_Up", ["in", "out", "up"]
                )
            elif control < position_target:
                total_layers[0][idx] = make_node(
                    tensors["cctrl_down"], "cCtrl_Down", ["in", "out", "up", "down"]
                )
            else:
                total_layers[0][idx] = make_node(
                    tensors["cctrl_up"], "cCtrl_Up", ["in", "out", "up", "down"]
                )

    return positions, total_layers
```

# TNCircuit class

This parent class stores the tensor layers and performs contractions. Key design:

**Storage**: `self.tensors` is `layers[time][qudit]` = Node or None. Each layer is a "time slice" of the circuit.

**Appending**: `append_operator` uses `append_operator_to_grid` to place gates at the earliest free time slot. Control wires (up/down) are connected when we contract.

**Contraction paths**:
- **time + statevector**: Contracts layer by layer through time into a single dense tensor. Exact but memory scales as 2^n.
- **space + statevector**: First contracts each qudit row across time, then folds the resulting row tensors from top to bottom. This shows that the same network can be reduced in a different order.
- **time + MPS**: Maintains the state as a product of small cores. After each gate, optionally compresses with `TT_State_Compression(eps)`. Memory scales with bond dimension.

**Immutability**: `_copy_layers()` returns a copy before any contraction. Queries never mutate the circuit.


```python
class TNCircuit:
    """
    Parent class for storing tensors and performing contractions.
    
    This class represents a quantum circuit as a tensor network, where:
    - Each qudit starts in the |0⟩ state
    - Gates are appended as tensor operations
    - The circuit can be contracted to compute the final state
    
    Attributes:
        tensors: A 2D list [time][qudit] storing tensor nodes at each time step
        n_qudits: Number of qudits in the circuit
        dimension_qudits: Dimension of each qudit (2 for qubits, higher for qudits)
        depth: Current depth (number of time layers) in the circuit
    """

    def __init__(self, n_qudits: int, qudits_dimension: int = 2):
        """
        Initialize circuit with all qudits in the |0⟩ state.
        
        Args:
            n_qudits: Number of qudits in the circuit
            qudits_dimension: Dimension of each qudit (default 2 for qubits)
            
        Raises:
            ValueError: If n_qudits or qudits_dimension is not positive
        """
        if n_qudits <= 0:
            raise ValueError(f"Number of qudits must be positive, got {n_qudits}.")
        if qudits_dimension <= 0:
            raise ValueError(
                f"Qudit dimension must be positive, got {qudits_dimension}."
            )

        # Create initial |0⟩ state: [1, 0, 0, ...] for each qudit
        initial_tensor = torch.zeros(qudits_dimension, dtype=DTYPE, device=device)
        initial_tensor[0] = 1

        # Initialize first layer with |0⟩ kets for all qudits
        self.tensors = [[
            make_node(initial_tensor.clone(), f"|0>_{qudit}", ["out"])
            for qudit in range(n_qudits)
        ]]
        self.n_qudits = n_qudits
        self.dimension_qudits = qudits_dimension
        self.depth = 1

    def _copy_layers(self):
        """
        Return a deep copy of the circuit layers.
        
        This ensures that contraction operations don't mutate the original circuit,
        allowing multiple queries without side effects.
        
        Returns:
            A deep copy of self.tensors
        """
        return copy_node_grid(self.tensors)

    def append_operator(self, operator: list, qudits: list):
        """
        Append a gate operator to the circuit at the specified qudits.
        
        The operator is placed at the earliest available time slot where all
        target qudits are free. Control wires are connected during contraction.
        
        Args:
            operator: List of tensor nodes representing the gate
            qudits: List of qudit indices where the gate should be applied
        """
        # Make a copy to avoid mutating the input operator
        operator_copy = copy_node_grid(operator)
        
        # Add the operator to the grid at the appropriate time layer
        append_operator_to_grid(
            self.tensors, operator_copy, qudits, n_qudits=self.n_qudits
        )
        
        # Update circuit depth
        self.depth = len(self.tensors)


    def _prepare_dense_layers(self, layers):
        """Connect dense-circuit edges and return the final output edges."""
        return connect_circuit_layers(layers, n_qudits=self.n_qudits)

    def _contract_statevector(self, layers):
        """
        Contract the circuit to produce a dense statevector.

        This method contracts all tensors along the time axis, producing the
        exact quantum state. Memory scales as O(d^n) where d is qudit dimension
        and n is the number of qudits.

        Args:
            layers: The tensor network layers to contract

        Returns:
            A single tensor node representing the final statevector

        Raises:
            ValueError: If the circuit structure is invalid (missing connections)
        """
        output_edges = self._prepare_dense_layers(layers)

        # Second pass: Contract all tensors
        # Start by contracting the initial layer across all qudits
        current_tensor = layers[0][0]
        for qudit in range(1, self.n_qudits):
            current_tensor = contract_between(
                current_tensor,
                layers[0][qudit],
                name=f"state^{qudit}_0",
                allow_outer_product=True,
            )

        # Contract each subsequent time layer
        for time_idx in range(1, len(layers)):
            for qudit, node in enumerate(layers[time_idx]):
                if node is not None:
                    current_tensor = contract_between(
                        current_tensor,
                        node,
                        name=f"state^{qudit}_{time_idx}",
                    )

        return reorder_dense_output_axes(current_tensor, output_edges)

    def _contract_space_statevector(self, layers):
        """
        Contract the circuit row by row and fold the result from top to bottom.

        Each qudit row is first contracted across time. The resulting row tensors
        are then combined using the vertical control bonds that happen to be open
        between neighbouring rows. Any missing vertical bond naturally becomes an
        outer product, which is why `allow_outer_product=True` is required here.
        """
        output_edges = self._prepare_dense_layers(layers)

        current_tensor = None
        for qudit in range(self.n_qudits):
            row_tensor = contract_grid_row(
                [layers[time_idx][qudit] for time_idx in range(len(layers))],
                name=f"space^{qudit}",
            )
            if current_tensor is None:
                current_tensor = row_tensor
            else:
                current_tensor = contract_between(
                    current_tensor,
                    row_tensor,
                    name=f"space^{qudit}",
                    allow_outer_product=True,
                )

        return reorder_dense_output_axes(current_tensor, output_edges)

    def _initialize_mps(self, initial_layer):
        """
        Convert the initial |0...0⟩ layer into canonical MPS cores.
        
        Each |0⟩ ket is converted to an MPS node with proper bond dimensions.
        
        Args:
            initial_layer: List of initial state nodes
            
        Returns:
            List of canonicalized MPS nodes
        """
        return [
            canonicalize_mps_node(node, name=f"MPS^{qudit}_0")
            for qudit, node in enumerate(initial_layer)
        ]

    def _contract_mps(self, layers, eps: float):
        """
        Evolve the MPS through each gate with optional compression.
        
        This method maintains the state as a Matrix Product State (MPS), which
        scales memory as O(n * d * χ²) where χ is the bond dimension. After each
        layer, the MPS can be compressed using TT decomposition to control χ.
        
        Args:
            layers: The tensor network layers to contract
            eps: Compression threshold (0 = no compression)
            
        Returns:
            List of MPS nodes representing the final state
        """
        # Initialize MPS from the |0...0⟩ state
        current_mps = self._initialize_mps(layers[0])

        # Apply each gate layer
        for time_idx in range(1, len(layers)):
            for qudit, node in enumerate(layers[time_idx]):
                if node is None:
                    continue

                # Connect gate input to current MPS output
                connect_edges(node["in"], current_mps[qudit]["out"])
                
                # Prepare axis names for the contracted tensor
                gate_axes = [
                    f"{axis_name}1"
                    for axis_name in node.axes_names
                    if axis_name in {"up", "down"}
                ]
                mps_axes = [
                    f"{axis_name}2"
                    for axis_name in current_mps[qudit].axes_names
                    if axis_name in {"up", "down"}
                ]
                has_out = "out" in node.axes_names
                axis_names = (["out"] if has_out else []) + gate_axes + mps_axes

                # Contract gate with MPS core
                contracted = contract_between(
                    node,
                    current_mps[qudit],
                    name=f"state^{qudit}_{time_idx}",
                    axis_names=axis_names,
                )

                # Process bond axes to maintain MPS structure
                bond_axes = axis_names[1:] if has_out else axis_names
                if bond_axes:
                    tensor_node, final_axis_names = process_axes(
                        contracted.tensor, bond_axes, has_out
                    )
                else:
                    tensor_node = contracted.tensor
                    final_axis_names = ["out"] if has_out else []

                # Canonicalize the updated MPS core
                current_mps[qudit] = canonicalize_mps_node(
                    make_node(tensor_node, f"MPS^{qudit}_{time_idx}", final_axis_names),
                    name=f"MPS^{qudit}_{time_idx}",
                )

            # Optionally compress the MPS after each layer
            if eps > 0:
                current_mps = TT_State_Compression(
                    current_mps, eps, device=device, dtype=DTYPE
                )

        return current_mps

    def _contract_layers(
        self, layers, representation: str = "statevector", eps: float = 0
    ):
        """
        Contract circuit layers using the specified representation.
        
        Args:
            layers: The tensor network layers to contract
            representation: Either "statevector" (exact, dense) or "mps" (compressed)
            eps: Compression threshold for MPS (ignored for statevector)
            
        Returns:
            Contracted state in the requested representation
            
        Raises:
            ValueError: If eps is negative or representation is unsupported
        """
        if eps < 0:
            raise ValueError("eps must be greater than or equal to 0.")

        representation_key = representation.lower()
        if representation_key == "statevector":
            return self._contract_statevector(layers)
        if representation_key == "mps":
            return self._contract_mps(layers, eps)
        raise ValueError("Supported representations are 'statevector' and 'MPS'.")


    def contract(
        self, scheme: str = "time", representation: str = "statevector", eps: float = 0
    ):
        """
        Contract the circuit to compute the final quantum state.

        This is the main public interface for querying the circuit state.
        The circuit itself is never mutated.

        Args:
            scheme: Contraction scheme ("time" or "space")
            representation: "statevector" for exact dense state, "mps" for compressed
            eps: Compression threshold for MPS (0 = no compression)

        Returns:
            The contracted quantum state in the requested representation

        Raises:
            NotImplementedError: If the scheme/representation pair is unsupported
        """
        if eps < 0:
            raise ValueError("eps must be greater than or equal to 0.")

        scheme_key = scheme.lower()
        representation_key = representation.lower()

        if scheme_key == "time":
            return self._contract_layers(
                self._copy_layers(), representation=representation, eps=eps
            )
        if scheme_key == "space":
            if representation_key != "statevector":
                raise NotImplementedError(
                    "Space contraction is only implemented for the statevector representation."
                )
            return self._contract_space_statevector(self._copy_layers())
        raise ValueError("Supported schemes are 'time' and 'space'.")
```

# QuantumRegister class

`QuantumRegister` maps qubit indices to human-readable labels (e.g. `q(0)`, `q(1)`).
Supports indexing with:
- `reg[0]` → single qudit
- `reg[0, 3]` → slice (qudits 0, 1, 2)
- `reg[[0, 2, 4]]` → list of positions
- `reg["all"]` → all qudits

Used by `QuantumCircuit` to resolve labels like `qr[0]` or `qr["all"]` to internal positions.


```python
class QuantumRegister:
    """A lightweight named register for qudit labels."""

    def __init__(self, n_qudits: int, name: str):
        if n_qudits <= 0:
            raise ValueError(f"Number of qudits must be positive, got {n_qudits}.")
        if not name or not isinstance(name, str):
            raise ValueError("Register name must be a non-empty string.")

        self.name = name
        self.n_qudits = n_qudits
        self.register_dict = {
            qudit: f"{name}({qudit})" for qudit in range(n_qudits)
        }

    def __call__(self) -> dict:
        return self.register_dict

    def __getitem__(self, items):
        if isinstance(items, list):
            try:
                return [self.register_dict[idx] for idx in items]
            except KeyError as exc:
                raise KeyError(
                    f"Qudit index {exc} out of range [0, {self.n_qudits})."
                ) from exc

        if items == "all":
            return [self.register_dict[idx] for idx in range(self.n_qudits)]

        if isinstance(items, tuple):
            if len(items) != 2:
                raise ValueError(
                    f"Tuple must have exactly 2 elements, got {len(items)}."
                )
            start, end = items
            if not (0 <= start < end <= self.n_qudits):
                raise ValueError(
                    f"Invalid range ({start}, {end}) for register with {self.n_qudits} qudits."
                )
            return [self.register_dict[idx] for idx in range(start, end)]

        try:
            return self.register_dict[items]
        except KeyError as exc:
            raise KeyError(
                f"Qudit index {items} out of range [0, {self.n_qudits})."
            ) from exc

    def __len__(self) -> int:
        return self.n_qudits
```

# QuantumCircuit class

This class inherits from `TNCircuit`, adds the Qiskit-like interface, and keeps the main public API intact. The internal implementation is smaller because it delegates low-level tensor work to the helpers defined above.


```python
class QuantumCircuit(TNCircuit):
    """Quantum circuit implementation using tensor networks.
    
    This class provides a Qiskit-like interface for building and manipulating
    quantum circuits using tensor network representations. It supports quantum
    registers, various quantum gates, and state vector computations.
    
    Args:
        *regs: Either a single integer specifying the number of qubits, or one
               or more QuantumRegister objects.
        name: Optional name for the circuit (default: "Circuit").
    
    Raises:
        ValueError: If no registers are provided or if mixing int with QuantumRegister.
        TypeError: If registers are not QuantumRegister instances.
    """

    def __init__(self, *regs, name: str = "Circuit"):
        if not regs:
            raise ValueError(
                "QuantumCircuit expects either a qubit count or one or more QuantumRegister objects."
            )

        self.name = name
        # Maps qudit labels (from registers) to their physical positions
        self.qudit_dict = {}

        # Handle initialization with integer (number of qubits)
        if isinstance(regs[0], int):
            if len(regs) != 1:
                raise ValueError(
                    "Pass either a single integer or one/more QuantumRegister objects."
                )
            n_qudits = regs[0]
            # For integer initialization, use simple index mapping
            self.qudit_dict = {index: index for index in range(n_qudits)}
        else:
            # Handle initialization with QuantumRegister objects
            n_qudits = 0
            for reg in regs:
                if not isinstance(reg, QuantumRegister):
                    raise TypeError(
                        "Registers must be QuantumRegister instances."
                    )
                # Map each qudit label in the register to its physical position
                for local_index in range(len(reg)):
                    self.qudit_dict[reg[local_index]] = n_qudits
                    n_qudits += 1

        # Initialize the parent TNCircuit with qubits (dimension=2)
        super().__init__(n_qudits, qudits_dimension=2)

    def __len__(self) -> int:
        """Return the depth (number of layers) of the circuit."""
        return self.depth

    def _resolve_position(self, qudit):
        """Convert a qudit identifier (string label or int index) to physical position.
        
        Args:
            qudit: Either a string label from a register or an integer index.
            
        Returns:
            int: The physical position of the qudit in the circuit.
            
        Raises:
            KeyError: If the qudit label/index is not found or out of range.
            TypeError: If the qudit identifier is neither string nor int.
        """
        # Handle string labels (from registers)
        if isinstance(qudit, str):
            if qudit not in self.qudit_dict:
                raise KeyError(f"Unknown qudit label: {qudit}")
            return self.qudit_dict[qudit]

        # Handle integer indices
        if not isinstance(qudit, int):
            raise TypeError(
                "Qudit identifiers must be integers or register labels."
            )

        if not 0 <= qudit < self.n_qudits:
            raise KeyError(f"Qudit index {qudit} out of range [0, {self.n_qudits}).")
        return qudit

    def _resolve_positions(self, qudits):
        """Convert multiple qudit identifiers to their physical positions.
        
        Args:
            qudits: A single qudit identifier or a list of identifiers.
            
        Returns:
            list: Physical positions corresponding to the input qudits.
        """
        # Wrap single qudit in a list for uniform processing
        if isinstance(qudits, (int, str)):
            qudits = [qudits]
        return [self._resolve_position(qudit) for qudit in qudits]

    def __getitem__(self, key):
        """Get all gate tensors applied to a specific qudit across all layers.
        
        Args:
            key: Qudit identifier (string label or int index).
            
        Returns:
            list: Gate tensors for the specified qudit at each circuit layer.
        """
        position = self._resolve_position(key)
        return [self.tensors[layer][position] for layer in range(self.depth)]

    def _append_single_qubit_gate(self, tensor, name: str, qudits) -> None:
        """Apply a single-qubit gate to one or more qudits.
        
        Args:
            tensor: The gate tensor (2x2 matrix for qubits).
            name: Name of the gate for visualization.
            qudits: Target qudit(s) - can be a single identifier or list.
        """
        positions = self._resolve_positions(qudits)
        # Create a layer with one gate node per target qudit
        layer = [[make_node(tensor, name, ["in", "out"]) for _ in positions]]
        self.append_operator(layer, positions)

    def x(self, qudits):
        """Apply Pauli-X (NOT) gate to specified qudit(s)."""
        self._append_single_qubit_gate(X_GATE, "X", qudits)

    def h(self, qudits):
        """Apply Hadamard gate to specified qudit(s)."""
        self._append_single_qubit_gate(H_GATE, "H", qudits)

    def z(self, qudits):
        """Apply Pauli-Z gate to specified qudit(s)."""
        self._append_single_qubit_gate(Z_GATE, "Z", qudits)

    def y(self, qudits):
        """Apply Pauli-Y gate to specified qudit(s)."""
        self._append_single_qubit_gate(Y_GATE, "Y", qudits)

    def rx(self, theta, qudits):
        """Apply rotation around X-axis by angle theta to specified qudit(s)."""
        tensor, name = rx_gate(theta)
        self._append_single_qubit_gate(tensor, name, qudits)

    def ry(self, theta, qudits):
        """Apply rotation around Y-axis by angle theta to specified qudit(s)."""
        tensor, name = ry_gate(theta)
        self._append_single_qubit_gate(tensor, name, qudits)

    def rz(self, theta, qudits):
        """Apply rotation around Z-axis by angle theta to specified qudit(s)."""
        tensor, name = rz_gate(theta)
        self._append_single_qubit_gate(tensor, name, qudits)

    def append_gate_control(self, tensor, name, positions_control, positions_target):
        """Apply a controlled gate with one or more controls and targets.
        
        Args:
            tensor: The gate tensor to be controlled.
            name: Name of the controlled gate.
            positions_control: Physical position(s) of control qudit(s).
            positions_target: Physical position(s) of target qudit(s).
        """
        positions, layers = build_control_gate_layers(
            tensor, name, positions_control, positions_target
        )
        self.append_operator(layers, positions)

    def cx(self, qudit_control, qudit_target):
        """Apply controlled-X (CNOT) gate.
        
        Args:
            qudit_control: Control qudit identifier(s).
            qudit_target: Target qudit identifier(s).
        """
        positions_control = self._resolve_positions(qudit_control)
        positions_target = self._resolve_positions(qudit_target)
        self.append_gate_control(X_GATE, "C-X", positions_control, positions_target)

    def cz(self, qudit_control, qudit_target):
        """Apply controlled-Z gate.
        
        Args:
            qudit_control: Control qudit identifier(s).
            qudit_target: Target qudit identifier(s).
        """
        positions_control = self._resolve_positions(qudit_control)
        positions_target = self._resolve_positions(qudit_target)
        self.append_gate_control(Z_GATE, "C-Z", positions_control, positions_target)

    def append_gate_multi_control(self, tensor, name, positions_control, positions_target):
        """Apply a gate with multiple control qudits and a single target.
        
        Args:
            tensor: The gate tensor to be controlled.
            name: Name of the multi-controlled gate.
            positions_control: List of physical positions of control qudits.
            positions_target: Physical position of the target qudit (or single-element list).
            
        Raises:
            ValueError: If positions_target contains more than one qudit.
        """
        # Ensure target is a single position, not a list
        if isinstance(positions_target, list):
            if len(positions_target) != 1:
                raise ValueError(
                    "Multi-controlled gates support exactly one target qudit."
                )
            positions_target = positions_target[0]

        positions, layers = build_multi_control_gate_layer(
            tensor, name, positions_control, positions_target
        )
        self.append_operator(layers, positions)

    def ccx(self, qudit_control1, qudit_control2, qudit_target):
        """Apply Toffoli (CCNOT) gate with two controls and one target.
        
        Args:
            qudit_control1: First control qudit identifier.
            qudit_control2: Second control qudit identifier.
            qudit_target: Target qudit identifier.
        """
        positions_control1 = self._resolve_position(qudit_control1)
        positions_control2 = self._resolve_position(qudit_control2)
        positions_target = self._resolve_position(qudit_target)
        self.append_gate_multi_control(
            X_GATE,
            "CC-X",
            [positions_control1, positions_control2],
            positions_target,
        )

    def state_vector(self) -> np.ndarray:
        """Compute and return the final state vector of the circuit.
        
        Returns:
            np.ndarray: The state vector as a 1D numpy array of complex amplitudes.
        """
        # Contract the tensor network using time-evolution scheme
        result = self.contract(scheme="time", representation="statevector", eps=0)
        return tensor_to_numpy_flat(result.tensor)

    def expected(
        self, operator, representation: str = "statevector", eps: float = 0
    ) -> complex:
        """Compute the expectation value <ψ|O|ψ> for a given operator.
        
        Args:
            operator: The operator as a grid of tensor nodes.
            representation: Contraction representation ("statevector" or "mps").
            eps: Truncation threshold for approximate contractions.
            
        Returns:
            complex: The expectation value.
            
        Raises:
            ValueError: If operator dimensions don't match circuit dimensions.
        """
        operator_copy = copy_node_grid(operator)
        # Verify operator has correct dimensions
        if any(len(layer) != self.n_qudits for layer in operator_copy):
            raise ValueError(
                "Operator must contain one tensor per circuit qudit in each layer."
            )

        # Build the full contraction: |ψ⟩ → O → ⟨ψ|
        layers = self._copy_layers()
        full_width = list(range(self.n_qudits))
        
        # Append the operator O
        append_operator_to_grid(
            layers, operator_copy, full_width, n_qudits=self.n_qudits
        )
        
        # Append the adjoint circuit ⟨ψ|
        append_operator_to_grid(
            layers,
            build_adjoint_grid(self.tensors),
            full_width,
            n_qudits=self.n_qudits,
        )

        # Contract and extract scalar result
        result = self._contract_layers(layers, representation, eps)
        return contraction_to_scalar(result, representation)

    def check_state(
        self, state: int, representation: str = "statevector", eps: float = 0
    ) -> complex:
        """Compute the amplitude of a specific computational basis state.
        
        Args:
            state: Integer representation of the basis state (e.g., 5 = |101⟩).
            representation: Contraction representation ("statevector" or "mps").
            eps: Truncation threshold for approximate contractions.
            
        Returns:
            complex: The amplitude ⟨state|ψ⟩.
            
        Raises:
            ValueError: If state is out of valid range.
        """
        if not 0 <= state < 2 ** self.n_qudits:
            raise ValueError(
                f"State must be in [0, {2 ** self.n_qudits}), got {state}."
            )

        # Convert state integer to bitstring (reversed for qubit ordering)
        bitstring = format(state, f"0{self.n_qudits}b")[::-1]
        
        # Create bra layer ⟨bitstring|
        bra_layer = build_basis_bra_layer(
            bitstring, device=device, dtype=DTYPE
        )

        # Contract circuit with bra to get amplitude
        layers = self._copy_layers()
        append_operator_to_grid(
            layers,
            bra_layer,
            list(range(self.n_qudits)),
            n_qudits=self.n_qudits,
        )

        result = self._contract_layers(layers, representation, eps)
        return contraction_to_scalar(result, representation)

    def to_gate(self) -> list:
        """Convert the circuit to a gate that can be used as a subcircuit.
        
        Returns:
            list: A copy of the circuit's tensor grid (excluding initial state layer).
        """
        return copy_node_grid(self.tensors[1:])

    def __repr__(self):
        """Return a string representation (ASCII diagram) of the circuit."""
        return draw_circuit(self.tensors, self.depth, self.n_qudits)
```

# Function that creates the Deutsch-Jozsa circuit

This function creates the Deutsch-Jozsa circuit for `n` input qubits in `f(x)` and a given oracle gate.


```python
def Deutsch_Jozsa_circuit(n_input_qubits: int, oracle) -> QuantumCircuit:
    """Create the Deutsch-Jozsa circuit for a given oracle."""

    qx = QuantumRegister(n_input_qubits, name="qx")
    qy = QuantumRegister(1, name="qy")

    qc = QuantumCircuit(qx, qy, name="Deutsch")

    qc.h(qx["all"])
    qc.x(qy[0])
    qc.h(qy[0])

    oracle_qudits = [
        qc.qudit_dict[qudit_id] for qudit_id in qx["all"] + [qy[0]]
    ]
    qc.append_operator(oracle, oracle_qudits)

    qc.h(qx["all"])
    return qc
```

# Function that executes the Deutsch-Jozsa algorithm

The algorithm contracts the circuit, checks the amplitude of the `00...0` state in the `x` register for both `0` and `1` in the `y` register, and combines them. If the total amplitude is approximately zero, the function is balanced; otherwise it is constant.


```python
def Deutsch_Jozsa_Algorithm(
    n_input_qubits: int, oracle, repr: str = "MPS", eps: float = 0
):
    """Run the Deutsch-Jozsa algorithm and print the classification."""

    qc = Deutsch_Jozsa_circuit(n_input_qubits, oracle)
    amp0 = qc.check_state(int("0" * (n_input_qubits + 1), 2), repr, eps)

    qc = Deutsch_Jozsa_circuit(n_input_qubits, oracle)
    amp1 = qc.check_state(int("1" + "0" * n_input_qubits, 2), repr, eps)

    total_amplitude = np.sqrt(np.abs(amp0) ** 2 + np.abs(amp1) ** 2)

    if np.round(total_amplitude, 6) == 0:
        print("Balanced function")
    else:
        print("Constant function")

    print(f"Amplitude of |0...0> state: {total_amplitude}")
```
