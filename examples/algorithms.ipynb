# Algorithms

Some canonical algorithms have been implemented to easily test them. Check out the following Deustch-Josza implementation.

```python
from qiskit_from_scratch import QuantumCircuit, Deustch_Jozsa_Algorithm

oracle = QuantumCircuit(3, name="Oracle")
# Balanced
oracle.cx(1, 2)

oracle_gate = oracle.to_gate()

Deustch_Jozsa_Algorithm(2, oracle_gate, repr="MPS", eps=0.5)
```

Constant

```python
oracle = QuantumCircuit(3, name="Oracle")
# Constant
oracle.x(2)

oracle_gate = oracle.to_gate()

Deustch_Jozsa_Algorithm(2, oracle_gate, repr="MPS", eps=0.3)
```
